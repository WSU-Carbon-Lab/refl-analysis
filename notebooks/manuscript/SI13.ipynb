# SI 13. ZnPc unit cells

The three unit cells used in this work, drawn to one length scale, each viewed along *b* (herringbone, *a* vertical, the film normal for α) and, beside it, one π stack edge-on (*b* vertical) with the stacking angle marked.

- **a** β-ZnPc, single-crystal structure at 100 K (`md/znpc_film/ff/tpZnPc_0m.cif`). Source of the rigid molecular frame and the force-field validation target.
- **b** α-ZnPc constrained cell. α-CuPc C2/c lattice (monoclinic angle 90.4°), *a* and *c* fixed by the GIXRD lamellar and (002) spacings, *b* by the π–π spacing, tilt from the lattice-energy scan (`cell/alpha_tilt_scan_73_79_86_90.json`, entry 0).
- **c** α-ZnPc relaxed cell. The constrained cell after rigid-body NPT at 300 K (`bulk/relaxed_cell_g73L.json`), four molecular sites averaged over the bulk simulation; the cell used to build the film.

Drawing and geometry live in `md/znpc_film/cell/unit_cell_figure.py`; this notebook only calls it.
